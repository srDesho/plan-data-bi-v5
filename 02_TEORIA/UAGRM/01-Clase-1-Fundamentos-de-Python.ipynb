### 💼 Caso de negocio: detectar al primer cliente en mora

## 12. Funciones: empaquetar lógica

Una función es un bloque de código con nombre. Lo defines una vez y
lo reutilizas todas las veces que necesites.

```python
def cuadrado(x):
    return x ** 2

print(cuadrado(5))
print(cuadrado(12))
```

```python
# Argumento con valor por defecto
def aplicar_descuento(monto, pct=0.10):
    return monto * (1 - pct)

print(aplicar_descuento(1000))         # 900.0 → usa 10% por defecto
print(aplicar_descuento(1000, 0.25))   # 750.0 → 25%
```

```python
# Función con docstring (la documentación interna)
def precio_final(precio, descuento=0.0, iva=0.13):
    '''Calcula el precio final después de descuento e IVA.'''
    return precio * (1 - descuento) * (1 + iva)

print(precio_final(1000))
print(precio_final(1000, descuento=0.10))
print(precio_final(1000, descuento=0.10, iva=0.00))
```

```python
# Devolver varios valores
def resumen(valores):
    return min(valores), max(valores), sum(valores) / len(valores)

ventas = [120, 80, 450, 60, 230, 1100]
mn, mx, prom = resumen(ventas)
print(f"min={mn}  max={mx}  promedio={prom:.1f}")
```

### 💼 Caso de negocio: score de cliente

Combinamos funciones para asignar un **score de 0 a 100** según consumo
y antigüedad, y luego clasificar al cliente.

```python
def score_cliente(consumo, meses, en_mora=False):
    '''Devuelve un score 0-100 según consumo (60%) y antigüedad (40%).'''
    if en_mora:
        return 0
    pts_consumo = min(consumo / 100, 60)   # hasta 60 puntos por consumo
    pts_meses   = min(meses * 2, 40)       # hasta 40 puntos por antigüedad
    return round(pts_consumo + pts_meses, 1)

def clasificar(score):
    if score >= 80: return "VIP"
    if score >= 40: return "Recurrente"
    if score >  0:  return "Nuevo"
    return "En riesgo"

# Aplicar a varios clientes
clientes_prueba = [
    {"id": 1, "consumo": 8200, "meses": 28, "en_mora": False},
    {"id": 2, "consumo": 1500, "meses":  4, "en_mora": False},
    {"id": 3, "consumo": 4500, "meses": 18, "en_mora": False},
    {"id": 4, "consumo": 7000, "meses": 14, "en_mora": True},
]

for c in clientes_prueba:
    s = score_cliente(c["consumo"], c["meses"], c["en_mora"])
    print(f"Cliente {c['id']}:  score={s:>5}   segmento={clasificar(s)}")
```

## 13. El puente al análisis de datos: NumPy

NumPy es **la** librería para cálculos numéricos en Python. Sus arrays
parecen listas, pero permiten operaciones sobre todos los elementos
a la vez (**vectorización**) — algo que con listas normales no se puede.

```python
import numpy as np

# Crear arrays
arpu = np.array([45, 60, 85, 30, 120, 95, 50, 200, 75, 40])
print(arpu)
print("tipo:", type(arpu), " | shape:", arpu.shape)
```

```python
# Operaciones vectorizadas → sobre todos los elementos a la vez
print(arpu * 1.13)        # aplicar IVA a todos
print(arpu - arpu.mean()) # diferencia con el promedio
```

```python
# Estadísticas básicas
print(f"ARPU promedio:  {arpu.mean():.2f} Bs")
print(f"ARPU mediano:   {np.median(arpu):.2f} Bs")
print(f"ARPU máximo:    {arpu.max()}")
print(f"ARPU mínimo:    {arpu.min()}")
print(f"Desviación est: {arpu.std():.2f}")
```

### 💼 Caso de negocio: análisis combinado de ARPU + churn

Combinamos dos arrays: el ARPU por usuario y un indicador de si el
usuario sigue activo. Calculamos los KPIs clave del mes.

```python
arpu        = np.array([45, 60, 85, 30, 120, 95, 50, 200, 75, 40])
esta_activo = np.array([1,  1,  1,  0,  1,   1,  0,  1,   1,  0])
# 1 = activo, 0 = se fue (churn)

# KPIs
arpu_global   = arpu.mean()
arpu_activos  = arpu[esta_activo == 1].mean()
arpu_churn    = arpu[esta_activo == 0].mean()
tasa_churn    = (esta_activo == 0).sum() / len(esta_activo)

print(f"ARPU global:           {arpu_global:.2f} Bs")
print(f"ARPU de activos:       {arpu_activos:.2f} Bs")
print(f"ARPU de los que se fueron: {arpu_churn:.2f} Bs")
print(f"Tasa de churn:         {tasa_churn:.0%}")
```

```python
# Identificar el top cuartil (clientes más valiosos)
umbral_top25 = np.quantile(arpu, 0.75)
top_clientes = arpu[arpu >= umbral_top25]

print(f"Umbral del top 25%: {umbral_top25:.2f} Bs")
print(f"Clientes en el top: {top_clientes}")
print(f"ARPU promedio del top: {top_clientes.mean():.2f} Bs")
```

### Arrays 2D: ventas por sucursal y trimestre

- `segmento` (`VIP`, `Recurrente`, `Nuevo`)
- `descuento_sugerido` (`0.15`, `0.08`, `0.02` según segmento)

```python
def clasificar_cliente(consumo_mensual, meses_activo):
    score = min(consumo_mensual / 5, 70) + min(meses_activo * 2.5, 30)
    score = round(score, 1)

    if score >= 80:
        segmento, descuento = "VIP", 0.15
    elif score >= 40:
        segmento, descuento = "Recurrente", 0.08
    else:
        segmento, descuento = "Nuevo", 0.02

    return {
        "score": score,
        "segmento": segmento,
        "descuento_sugerido": descuento,
    }

# Probar
print(clasificar_cliente(consumo_mensual=350, meses_activo=24))
print(clasificar_cliente(consumo_mensual=80,  meses_activo=3))
print(clasificar_cliente(consumo_mensual=200, meses_activo=10))
```

### 🧩 Ejercicio 4 — Análisis de stock crítico

Tienes el stock actual de cada producto y el stock mínimo aceptable.
Identifica los productos que están **por debajo del mínimo**.

```python
stock_actual = {
    "PLAN-4G":   850,
    "PLAN-5G":    18,
    "CHIP":     1200,
    "ROUTER":     12,
    "AURICULAR": 180,
}

stock_minimo = {
    "PLAN-4G":   500,
    "PLAN-5G":    50,
    "CHIP":      800,
    "ROUTER":     30,
    "AURICULAR": 100,
}

# Tu solución
criticos = []
for producto, actual in stock_actual.items():
    minimo = stock_minimo[producto]
    if actual < minimo:
        criticos.append((producto, actual, minimo, minimo - actual))

print("⚠️  Productos en stock crítico:")
for p, act, mn, faltante in criticos:
    print(f"  {p:>10}: actual={act:>4}, mínimo={mn:>4}  → reponer {faltante}")
```

### 🧩 Ejercicio 5 — KPIs con NumPy

Tienes los ingresos mensuales del último año. Calcula:

1. Ingreso total y promedio mensual.
2. Mes con mayor ingreso (índice).
3. Variación porcentual mes a mes.
4. Cuántos meses estuvieron **por encima del promedio**.

```python
ingresos = np.array([
    12500, 13800, 11200, 14500, 16800, 15200,
    17900, 19500, 18200, 21000, 23500, 26800,
])

# 1
print(f"Total:    {ingresos.sum():,} Bs")
print(f"Promedio: {ingresos.mean():,.0f} Bs")

# 2
mes_max = ingresos.argmax()
print(f"Mejor mes: índice {mes_max} (mes {mes_max + 1}) con {ingresos[mes_max]:,} Bs")

# 3 — variación mes a mes
variacion = np.diff(ingresos) / ingresos[:-1]
print("Variación m/m:", np.round(variacion * 100, 1), "%")

# 4
sobre_promedio = (ingresos > ingresos.mean()).sum()
print(f"Meses sobre el promedio: {sobre_promedio}")
```
