
# Descripción de variables seleccionadas — UCI Automobile Dataset

| Variable | Descripción en español (público no técnico) |
|---------|---------------------------------------------|
| **aspiration** | Indica si el motor es normal (“std”) o turbo. Los motores turbo comprimen más aire y generan mayor potencia. |
| **body-style** | Tipo de carrocería del vehículo: sedán, hatchback, convertible, wagon, etc. |
| **bore** | Diámetro interno del cilindro del motor. Afecta la potencia que puede generar el motor. |
| **city-mpg** | Rendimiento de combustible en ciudad (millas por galón). Entre más alto, menor consumo. |
| **compression-ratio** | Relación de compresión dentro del motor; influye en eficiencia y rendimiento. |
| **curb-weight** | Peso del auto sin pasajeros ni carga, pero con combustible y fluidos. |
| **drive-wheels** | Tipo de tracción: delantera (**fwd**), trasera (**rwd**) o 4x4 (**4wd**). |
| **engine-location** | Ubicación del motor: delantero o trasero. |
| **engine-size** | Tamaño o capacidad del motor (volumen interno). Motores más grandes producen más potencia. |
| **engine-type** | Tipo de diseño del motor (DOHC, OHC, etc.). |
| **fuel-system** | Tipo de sistema de alimentación de combustible (inyección, carburador, etc.). |
| **fuel-type** | Tipo de combustible: gasolina o diésel. |
| **height** | Altura total del automóvil. |
| **highway-mpg** | Rendimiento de combustible en carretera. Suelen ser valores más altos que en ciudad. |
| **horsepower** | Potencia del motor medida en caballos de fuerza (HP). Determina la capacidad de aceleración. |
| **length** | Largo total del vehículo. |
| **make** | Marca del automóvil (Toyota, BMW, Honda, etc.). |
| **normalized-losses** | Indicador del promedio de pérdidas que paga el seguro por este tipo de auto, comparado con autos similares. |
| **num-of-cylinders** | Número de cilindros del motor (4, 6, 8, etc.). |
| **num-of-doors** | Número de puertas del vehículo (2 o 4), expresado en texto. |
| **peak-rpm** | Revoluciones máximas del motor por minuto. |
| **price** | Precio del vehículo en dólares. |
| **stroke** | Longitud del recorrido del pistón dentro del cilindro. Junto con *bore*, determina la capacidad del motor. |
| **wheel-base** | Distancia entre los ejes delantero y trasero. Influye en estabilidad y comodidad. |
| **width** | Ancho total del automóvil. |

Definicion del intervalo

image.png

```python
# necesitamos definir el maximo y minimo de precios

X.price.min(),X.price.max() # cuidado con los valores de maximo y minimo crudos, estos pueden ser OUTLIERS (observaciones anomalas, se explicara posteriormente)
```

```python
k = 8

intervalo = (X.price.max() - X.price.min())/k

intervalo
```

```python
import numpy as np
```

```python
# quiero crear 8 intervalos

np.linspace(5000,50000,9)
```

```python
# definicion de intervalos bajo teoria de construccion de intervalos

pd.cut(X.price, bins = np.linspace(5000,50000,9)).value_counts() # solamente estamos viendo la frecuencia absoluta
```

```python
# otro metodo sencillo de realizar este mismo ejercicio con una cantidad fija de datos por intervalo

pd.qcut(X.price, 8).value_counts()
```

```python
# el comando reset_index hace que el index se haga columna

df_brackets_price = pd.cut(X.price, bins = np.linspace(5000,50000,9)).value_counts().reset_index() # cuidado con las palabras reservadas
```

```python
df_brackets_price
```

```python
# error por palabra reservada

# df_brackets_price.count.sum()
```

```python
df_brackets_price.rename({'count':'n_obs'}, axis = 1, inplace=True)
```

```python
df_brackets_price
```

```python
df_brackets_price.n_obs.sum()
```

```python
# veamos la frecuencia relativa (proporcion) de los precios

df_brackets_price['f_relativa'] = df_brackets_price['n_obs']/df_brackets_price.n_obs.sum()
```
